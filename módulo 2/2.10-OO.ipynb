#### Q1.
Essa lista de exercícios terá como base a classe `Evento` criada em exercícios anteriores. Primeiramente criaremos a classe abstrata `EventoABC` com os métodos de instância abstratos `__str__(self)` e `isConcluido(self)`, indicando que todos as subclasses que dela herdarem devem implementar esses métodos.

`EventoABC` também possui os atributos `_titulo` (string) e `_descricao` (string), cujos valores são recebidos e inicializados no construtor da classe. Note a convenção de nomenclatura indicando o caráter privado desses atributos.

````python
from abc import ABC, abstractmethod

class EventoABC(ABC):
    """
    Classe abstrata que define a interface para eventos.
    """
    
    def __init__(self, titulo, descricao):
        """
        Construtor da classe EventoABC.
        
        Args:
            titulo (str): Título do evento
            descricao (str): Descrição do evento
        """
        self._titulo = titulo
        self._descricao = descricao
    
    @abstractmethod
    def __str__(self):
        """
        Método abstrato que deve ser implementado pelas subclasses.
        Retorna uma representação em string do evento.
        """
        pass
    
    @abstractmethod
    def isConcluido(self):
        """
        Método abstrato que deve ser implementado pelas subclasses.
        Verifica se o evento foi concluído.
        
        Returns:
            bool: True se o evento foi concluído, False caso contrário
        """
        pass
````

Crie a classe `DataHora` que dará suporte ao registro de eventos de calendário.

* A classe **não possui construtor customizado**. A alteração de seu atributo se dará a partir da propriedade a seguir.

    * O setter da propriedade deve receber uma string de data formatada (`%d/%m/%Y, %H:%M`) e implementar um bloco try-except que tenta converter a string em `datetime` e lança um `ValueError` caso a entrada seja inválida. Use o atributo `FORMAT`. Consulte o [funcionamento do método `strptime`](https://www.digitalocean.com/community/tutorials/python-string-to-datetime-strptime).
* Crie o método de instância `isPassado(self)` que avalia se a `_data_hora` é menor que `datetime.now()` (a data e hora atual) e retorna `True` em caso positivo, e `False` caso contrário.

````python
from datetime import datetime, timedelta

class DataHora:
    """
    Classe que gerencia datas e horas para eventos de calendário.
    """
    
    # Atributo de classe - formato padrão
    FORMAT = '%d/%m/%Y, %H:%M'
    
    def __init__(self):
        """
        Construtor sem parâmetros customizados.
        Inicializa _data_hora como None.
        """
        self._data_hora = None
    
    @property
    def data_hora(self):
        """
        Getter da propriedade data_hora.
        
        Returns:
            str: Data e hora formatada no padrão '%d/%m/%Y, %H:%M'
        """
        if self._data_hora is None:
            return None
        return self._data_hora.strftime(self.FORMAT)
    
    @data_hora.setter
    def data_hora(self, valor):
        """
        Setter da propriedade data_hora.
        
        Args:
            valor (str): String de data formatada '%d/%m/%Y, %H:%M'
            
        Raises:
            ValueError: Se a string não puder ser convertida para datetime
        """
        try:
            self._data_hora = datetime.strptime(valor, self.FORMAT)
        except ValueError as e:
            raise ValueError(f"Formato de data inválido. Use o formato: {self.FORMAT}") from e
    
    def isPassado(self):
        """
        Verifica se a data e hora é passada (anterior à data/hora atual).
        
        Returns:
            bool: True se a data é passada, False caso contrário
        """
        return self._data_hora < datetime.now()
    
    def somaDias(self, num_dias):
        """
        Soma dias à data e hora atual e retorna a string formatada.
        
        Args:
            num_dias (int): Número de dias a serem somados
            
        Returns:
            str: String formatada da data com os dias somados
        """
        data_hora_somada = self._data_hora + timedelta(days=num_dias)
        return data_hora_somada.strftime(self.FORMAT)


# ===== TESTE DA CLASSE =====

# instanciando o objeto
dh = DataHora()

# definindo a data_hora através da propriedade
dh.data_hora = '05/02/2024, 12:30'

# editando a data_hora através da função somaDias
dh.data_hora = dh.somaDias(30)

# imprimindo a data_hora editada e se é passado
print(dh.data_hora, dh.isPassado())
````

#### Q3.
Crie a classe `EventoUnico`:
* A classe deve herdar de `EventoABC`.
* Possui o atributo de instância `_data_hora` (classe `DataHora` que criamos previamente).
* Seu construtor deve receber e inicializar os atributos da superclasse, além do valor de `_data_hora` recebido como uma string formatada (`%d/%m/%Y, %H:%M`). Note que para alterar `_data_hora` (objeto tipo `DataHora`), você deve manipular a propriedade interna da classe.
*  Implementa os métodos abstratos da superclasse:
    * Método `isConcluido()` que invoca o método `isPassado()` de `_data_hora` e retorna o seu resultado.
    * Método `__str__` que imprime os atributos do evento na forma `"Evento: _titulo, Data: _data_hora, Descrição: _descricao, Concluido: isConcluido()"`. Note que `isConcluido()` é o método de avaliação implementado. 
* Crie o método de instância `editar_data_hora` que recebe uma string formatada e altera `_data_hora` (através de sua propriedade interna).
    
    
Teste a classe `EventoUnico` com o seguinte código:
```python

# editar data do evento (através da propriedade)
```

````python
class EventoUnico(EventoABC):
    """
    Classe que representa um evento que ocorre uma única vez.
    Herda de EventoABC e implementa seus métodos abstratos.
    """
    
    def __init__(self, titulo, descricao, data_hora_str):
        """
        Construtor da classe EventoUnico.
        
        Args:
            titulo (str): Título do evento
            descricao (str): Descrição do evento
            data_hora_str (str): Data e hora formatada '%d/%m/%Y, %H:%M'
        """
        # Inicializa atributos da superclasse
        super().__init__(titulo, descricao)
        
        # Cria objeto DataHora e atribui o valor através da propriedade
        self._data_hora = DataHora()
        self._data_hora.data_hora = data_hora_str
    
    def isConcluido(self):
        """
        Verifica se o evento foi concluído.
        Um evento é considerado concluído se sua data já passou.
        
        Returns:
            bool: True se a data passou, False caso contrário
        """
        return self._data_hora.isPassado()
    
    def __str__(self):
        """
        Representação em string do evento único.
        
        Returns:
            str: String formatada com informações do evento
        """
        return (f"Evento: {self._titulo}, Data: {self._data_hora.data_hora}, "
                f"Descrição: {self._descricao}, Concluido: {self.isConcluido()}")
    
    def editar_data_hora(self, nova_data_hora_str):
        """
        Edita a data e hora do evento.
        
        Args:
            nova_data_hora_str (str): Nova data formatada '%d/%m/%Y, %H:%M'
        """
        self._data_hora.data_hora = nova_data_hora_str


# ===== TESTE DA CLASSE =====

# criar evento
evento = EventoUnico('Reunião', 'Sala 302, prédio da esquina', '05/10/2023, 16:30')
print(evento)

# editar data do evento (através da propriedade)
evento.editar_data_hora('05/10/2024, 16:30')
print(evento)
````

#### Q3.
Crie a classe `EventoRecorrente`:
* A classe deve herdar de `EventoABC`.
* Possui como atributo próprio uma lista privada de objetos `DataHora` (como você deve nomear o atributo?).
* Seu construtor recebe os atributos da superclasse, além dos atributos `data_hora_inicial` (string formatada), `data_hora_final` (string formatada) e `intervalo_repeticao` (int), sendo o intervalo dado em dias. Preencha a coleção `DataHora` de acordo com o intervalo de repetição fornecido. Dica: crie o objeto `DataHora` inicial e use sua função interna `somaDias` para criar iterativamente as novas instâncias do intervalo até chegar em `DataHora` final. 
*  Implementa os métodos abstratos da superclasse:
    * Método `isConcluido(indice)` que que invoca o método `isPassado()` do elemento `indice` da coleção de objetos `DataHora` e retorna seu resultado. 
    * Método `__str__` que imprime (em um laço) **todos as ocorrências `i` do evento** na forma `"Evento: _titulo, Data: data_hora[i], Descrição: _descricao, Concluido: isConcluido(i)"`. 
* Crie o método `editar_data_hora` que recebe `data_hora_antiga` e `data_hora_nova` e altera o elemento da coleção de objetos `DataHora` que corresponde a `data_hora_antiga` fornecida.    


Teste a classe `EventoRecorrente` com o seguinte código:
```python
# criar evento
eventos = EventoRecorrente(
    'Reunião', 'Sala 302, prédio da esquina', 
    '05/01/2024, 16:30', '05/01/2025, 16:30', 30)

# imprimir eventos
print(eventos)

# editar um dos eventos
eventos.editar_data_hora('05/12/2024, 16:30', '05/12/2024, 11:30')

# imprimir eventos
print(eventos)
```

````python
class EventoRecorrente(EventoABC):
    """
    Classe que representa um evento que se repete em intervalos regulares.
    Herda de EventoABC e implementa seus métodos abstratos.
    """
    
    def __init__(self, titulo, descricao, data_hora_inicial, data_hora_final, intervalo_repeticao):
        """
        Construtor da classe EventoRecorrente.
        
        Args:
            titulo (str): Título do evento
            descricao (str): Descrição do evento
            data_hora_inicial (str): Data inicial formatada '%d/%m/%Y, %H:%M'
            data_hora_final (str): Data final formatada '%d/%m/%Y, %H:%M'
            intervalo_repeticao (int): Intervalo em dias entre as ocorrências
        """
        # Inicializa atributos da superclasse
        super().__init__(titulo, descricao)
        
        # Cria a lista privada de objetos DataHora
        self._data_horas = []
        
        # Cria o objeto DataHora inicial
        data_hora_atual = DataHora()
        data_hora_atual.data_hora = data_hora_inicial
        
        # Cria o objeto DataHora final para comparação
        data_hora_final_obj = DataHora()
        data_hora_final_obj.data_hora = data_hora_final
        
        # Preenche a lista com objetos DataHora em intervalos
        while data_hora_atual.data_hora <= data_hora_final:
            # Cria novo objeto DataHora e adiciona à lista
            nova_data_hora = DataHora()
            nova_data_hora.data_hora = data_hora_atual.data_hora
            self._data_horas.append(nova_data_hora)
            
            # Avança para o próximo intervalo
            proxima_data_str = data_hora_atual.somaDias(intervalo_repeticao)
            data_hora_atual.data_hora = proxima_data_str
    
    def isConcluido(self, indice):
        """
        Verifica se a ocorrência no índice fornecido já passou.
        
        Args:
            indice (int): Índice da ocorrência na lista
            
        Returns:
            bool: True se a data passou, False caso contrário
        """
        return self._data_horas[indice].isPassado()
    
    def __str__(self):
        """
        Representação em string do evento recorrente.
        Imprime todas as ocorrências do evento.
        
        Returns:
            str: String formatada com informações de todas as ocorrências
        """
        resultado = ""
        for i in range(len(self._data_horas)):
            resultado += (f"Evento: {self._titulo}, Data: {self._data_horas[i].data_hora}, "
                         f"Descrição: {self._descricao}, Concluido: {self.isConcluido(i)}\n")
        
        # Remove a última quebra de linha
        return resultado.rstrip('\n')
    
    def editar_data_hora(self, data_hora_antiga, data_hora_nova):
        """
        Edita a data e hora de uma ocorrência específica.
        
        Args:
            data_hora_antiga (str): Data antiga formatada '%d/%m/%Y, %H:%M'
            data_hora_nova (str): Nova data formatada '%d/%m/%Y, %H:%M'
            
        Raises:
            ValueError: Se a data antiga não for encontrada na lista
        """
        # Procura pela data antiga na lista
        for data_hora in self._data_horas:
            if data_hora.data_hora == data_hora_antiga:
                # Encontrou a data, altera para a nova
                data_hora.data_hora = data_hora_nova
                return
        
        # Se chegou aqui, a data não foi encontrada
        raise ValueError(f"Data {data_hora_antiga} não encontrada nas ocorrências do evento")


# ===== TESTE DA CLASSE =====

# criar evento
eventos = EventoRecorrente(
    'Reunião', 'Sala 302, prédio da esquina', 
    '05/01/2024, 16:30', '05/01/2025, 16:30', 30)

# imprimir eventos
print(eventos)
print()

# editar um dos eventos
eventos.editar_data_hora('05/12/2024, 16:30', '05/12/2024, 11:30')

# imprimir eventos
print(eventos)
````

#### Q4.

Por fim, vamos só ver o polimorfismo em ação. Crie e preencha uma lista de eventos, sendo alguns do tipo `EventoUnico` e outros do tipo `EventoRecorrente`. Sobre essa lista, execute o laço de impressão a seguir:
```python
for evento in lista_eventos: print(evento)
```
A função `print` irá invocar o método especial `__str__` das classes correspondentes dependendo do tipo do objeto recebido. Aí está o polimorfismo :)

````python
# ===== DEMONSTRAÇÃO DE POLIMORFISMO =====

# Criar uma lista de eventos misturando EventoUnico e EventoRecorrente
lista_eventos = []

# Adicionar alguns EventoUnico
evento1 = EventoUnico('Apresentação', 'Sala de conferências A', '15/04/2026, 10:00')
evento2 = EventoUnico('Almoço de trabalho', 'Restaurante próximo', '20/04/2026, 12:00')
evento3 = EventoUnico('Entrega de projeto', 'Via email', '10/04/2026, 17:00')

lista_eventos.append(evento1)
lista_eventos.append(evento2)
lista_eventos.append(evento3)

# Adicionar alguns EventoRecorrente
evento4 = EventoRecorrente(
    'Reunião semanal',
    'Sala 302, prédio da esquina',
    '01/04/2026, 09:00',
    '15/04/2026, 09:00',
    7  # intervalo de 7 dias
)

evento5 = EventoRecorrente(
    'Treinamento mensal',
    'Auditório principal',
    '05/04/2026, 14:00',
    '05/05/2026, 14:00',
    30  # intervalo de 30 dias
)

lista_eventos.append(evento4)
lista_eventos.append(evento5)

# ===== POLIMORFISMO EM AÇÃO =====
# O método __str__ da classe apropriada será chamado automaticamente

print("=" * 80)
print("DEMONSTRAÇÃO DE POLIMORFISMO - LISTAGEM DE EVENTOS")
print("=" * 80)
print()

for evento in lista_eventos:
    print(evento)
    print("-" * 80)
    print()
````
